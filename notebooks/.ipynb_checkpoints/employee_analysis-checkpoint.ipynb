```python
# Employee Data Analysis

## Introduction
This project analyzes employee data using Python and pandas.

## Objectives
- Explore the employee dataset.
- Calculate salary statistics.
- Compare salaries by department.
- Analyze employees by city.
- Measure execution time using %timeit.
    
```

```python

import pandas as pd

df = pd.read_csv("../data/employees.csv")

df.head()
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python

print("Average salary:", df["salary"].mean())
print("Minimum salary:", df["salary"].min())
print("Maximum salary:", df["salary"].max())

highest_paid = df.loc[df["salary"].idxmax()]

print("\nHighest-paid employee:")
print(highest_paid[["name", "department", "salary"]])
```

```python

print("Employees by department:")
print(df["department"].value_counts())

print("\nAverage salary by department:")
print(df.groupby("department")["salary"].mean())

print("\nEmployees by city:")
print(df["city"].value_counts())

print("\nAverage salary by city:")
print(df.groupby("city")["salary"].mean())
```

```python

%timeit df["salary"].mean()
```

```python

## 4. Key Findings

1. The dataset contains 8 employees and 5 columns.
2. The average salary is 58,500.
3. Karthik has the highest salary at 90,000.
4. IT has the largest number of employees, with 3.
5. Chennai has the largest number of employees, with 4.

## Conclusion

Pandas helps us explore employee information and
compare salaries across departments and cities.
The findings are based on the supplied sample dataset.
```
